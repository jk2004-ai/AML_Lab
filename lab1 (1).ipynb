```python
import pandas as pd


df = pd.read_csv("/train.csv")


print(df.columns)


#df.info()

# 1. Total number of passengers
total_number_of_passengers = len(df)
print("Total number of passengers:", total_number_of_passengers)

# 2. Average age of survivors and non-survivors
average_age = df.groupby("Survived")["Age"].mean()
print("Average age:")
print(average_age)

# 3. Passengers older than 60 years
older_than_60 = df[df["Age"] > 60]
print("Passengers older than 60 years:")
print(older_than_60)

# 4. Average fare for each passenger class
average_fare = df.groupby("Pclass")["Fare"].mean()
print("Average fare for each passenger class:")
print(average_fare)
```
