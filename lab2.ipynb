```python
from sklearn.datasets import load_iris
iris = load_iris(as_frame=True)
df = iris.frame
df.head()
```

```python
jk.scatter(df["sepal length (cm)"],
            df["sepal width (cm)"])
jk.xlabel("sepla length (cm)")
jk.ylabel("sepal width(cm)")
jk.show()
```

```python
jk.hist(df["petal width (cm)"],bins=10)
jk.title("histogr")

jk.show()
```

```python
species = ['setosa','versicolor','virginica']
count = df['target'].value_counts()
jk.bar(species,count)
jk.show()
```
